## ขั้นที่ 2: Profile ทุกคอลัมน์

### 2.1 โหลดข้อมูล

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_rows", 100)
pd.set_option("display.max_columns", 50)

df = pd.read_csv("../data/sample_data_intw.csv")
print("rows, columns =", df.shape)
df.drop(columns="msisdn").head()   # ไม่แสดงเบอร์โทร
```

### 2.2 ชนิดข้อมูล ค่าว่าง และค่าซ้ำ

```python
print(df.dtypes.value_counts(), "\n")
print("ค่าว่างทั้งหมด:", df.isna().sum().sum())
print("แถวที่ซ้ำกันทุกคอลัมน์:", df.duplicated().sum())
print("จำนวนเบอร์ (msisdn) ไม่ซ้ำ:", df["msisdn"].nunique(), "จาก", len(df), "แถว")
```

### 2.3 ตารางสรุปรายคอลัมน์

```python
num = df.select_dtypes("number")
profile = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "distinct": df.nunique(),
    "min": num.min(),
    "median": num.median(),
    "max": num.max(),
    "pct_zero": (num == 0).mean().mul(100).round(1),
    "n_negative": (num < 0).sum(),
}).reindex(df.columns)
profile
```

### 2.4 คอลัมน์ที่เป็นข้อความหรือวันที่

```python
print(df["pcircle"].value_counts(), "\n")
pdate = pd.to_datetime(df["pdate"], format="%d-%m-%Y")
print(pdate.min().date(), "ถึง", pdate.max().date(), "|", pdate.nunique(), "วัน")
```

### 2.5 เช็กเพิ่มเติม

- คอลัมน์ไหนมีค่าติดลบ
- `maxamnt_loans30`: คำอธิบายบอกว่ามีแค่ 5 กับ 10 จริงไหม
- เบอร์เดียวกันมีหลายแถวไหม และมีเบอร์ที่ได้ทั้ง label 0 และ 1 ไหม

```python
neg = profile.loc[profile["n_negative"] > 0, ["min", "n_negative"]]
print("คอลัมน์ที่มีค่าติดลบ:\n", neg, "\n")

print("maxamnt_loans30 ค่าที่พบบ่อย:\n", df["maxamnt_loans30"].value_counts().head(10), "\n")
print("maxamnt_loans30 ที่ไม่ใช่ 0, 5, 6, 10, 12:", (~df["maxamnt_loans30"].isin([0, 5, 6, 10, 12])).sum(), "แถว\n")

rep = df["msisdn"].value_counts()
print("เบอร์ที่มีมากกว่า 1 แถว:", (rep > 1).sum(), "เบอร์")
print("แถวที่เกินมาจากเบอร์ซ้ำ:", len(df) - df["msisdn"].nunique(), "แถว")
print("เบอร์ที่มีทั้ง label 0 และ 1:", (df.groupby("msisdn")["label"].nunique() > 1).sum(), "เบอร์")
```

### 2.6 ค่าที่น่าจะเสีย (corrupted)

คอลัมน์ที่เป็น "จำนวนครั้ง" หรือ "จำนวนวัน" ต้องเป็นจำนวนเต็ม แต่ในตาราง 2.3 มีค่าทศนิยมขนาดใหญ่เกือบ 100,000 หรือ 1,000,000 ปนอยู่ (เช่น `aon` max = 999,860.76)

```python
int_like = ["aon", "last_rech_date_ma", "last_rech_date_da", "cnt_ma_rech30", "fr_ma_rech30",
            "cnt_da_rech30", "fr_da_rech30", "cnt_loans90", "maxamnt_loans30"]
bad = pd.DataFrame({c: (df[c] % 1 != 0) for c in int_like})
# medianmarechprebal30 เป็นยอดเงิน (มีทศนิยมได้) จึงใช้เกณฑ์ >= 100,000 แทน ช่วง 90 วันสูงสุดแค่ 41,457
bad["medianmarechprebal30"] = df["medianmarechprebal30"] >= 100_000
print("จำนวนแถวที่เป็นทศนิยม (ไม่ควรเป็น) ต่อคอลัมน์:\n", bad.sum(), "\n")
print("ค่าต่ำสุดของค่าที่เป็นทศนิยมพวกนี้:", round(min(df.loc[bad[c], c].min() for c in bad.columns if bad[c].any()), 1))
print("แถวที่มีอย่างน้อย 1 ค่าเสีย:", bad.any(axis=1).sum(), "แถว")
```

### 2.7 ตรวจความสอดคล้องระหว่างคอลัมน์

ตั้งกฎจากความหมายของคอลัมน์ แล้วนับแถวที่ผิดกฎ
- ยอดคืนต่อครั้งมีแค่ 6 หรือ 12 ดังนั้น `maxamnt_loans90` ต้องเป็น 0/6/12 และ `amnt_loans` ต้องหาร 6 ลงตัว
- ช่วง 30 วันอยู่ในช่วง 90 วัน ดังนั้นค่าช่วง 30 วันต้องไม่มากกว่าช่วง 90 วัน
- ข้อมูลย้อนหลัง 90 วัน ดังนั้นจำนวนวันนับจากเติมเงินครั้งล่าสุดไม่ควรเกิน 90 (นับเฉพาะค่าจำนวนเต็ม เพราะค่าทศนิยมนับเป็นค่าเสียไปแล้วในข้อ 2.6)

```python
rules = {
    "maxamnt_loans90 ไม่ใช่ 0/6/12":           ~df["maxamnt_loans90"].isin([0, 6, 12]),
    "amnt_loans30 หาร 6 ไม่ลงตัว":             df["amnt_loans30"] % 6 != 0,
    "amnt_loans90 หาร 6 ไม่ลงตัว":             df["amnt_loans90"] % 6 != 0,
    "cnt_loans30 > cnt_loans90":               df["cnt_loans30"] > df["cnt_loans90"],
    "amnt_loans30 > amnt_loans90":             df["amnt_loans30"] > df["amnt_loans90"],
    "cnt_ma_rech30 > cnt_ma_rech90":           df["cnt_ma_rech30"] > df["cnt_ma_rech90"],
    "sumamnt_ma_rech30 > sumamnt_ma_rech90":   df["sumamnt_ma_rech30"] > df["sumamnt_ma_rech90"],
    "last_rech_date_ma > 90 วัน (จำนวนเต็ม)":  (df["last_rech_date_ma"] > 90) & (df["last_rech_date_ma"] % 1 == 0),
    "last_rech_date_da > 90 วัน (จำนวนเต็ม)":  (df["last_rech_date_da"] > 90) & (df["last_rech_date_da"] % 1 == 0),
}
pd.Series({k: int(v.sum()) for k, v in rules.items()}, name="แถวที่ผิดกฎ").to_frame()
```

## ขั้นที่ 3: ยืนยัน target (`label`)

คำอธิบายจากเจ้าของข้อมูล: `label` = คืนเงินกู้ภายใน 5 วันหรือไม่ (1 = คืน = ลูกค้าดี, 0 = ไม่คืน = ลูกค้าไม่ดี)

ต้องพิสูจน์ด้วยข้อมูลจริง 2 เรื่อง
1. `label` มีแค่ค่า 0 กับ 1 จริงไหม
2. ทิศทางถูกไหม ถ้า 0 คือลูกค้าไม่ดี กลุ่ม 0 ควรใช้จ่ายน้อยกว่า เติมเงินน้อยกว่า และมีเงินน้อยกว่ากลุ่ม 1

### 3.1 `label` มีค่าอะไรบ้าง

```python
print(df["label"].value_counts().sort_index())
```

### 3.2 เทียบพฤติกรรมของกลุ่ม 0 กับกลุ่ม 1

ใช้ median แทน mean เพราะมีค่าเสียหลักแสนหลักล้านปนอยู่ ซึ่งจะดึงค่าเฉลี่ยให้เพี้ยน

```python
cols = ["daily_decr30", "cnt_ma_rech30", "sumamnt_ma_rech30", "medianmarechprebal30", "aon", "cnt_loans30"]
compare = df.groupby("label")[cols].median().T
compare.columns = ["label 0", "label 1"]
compare["กลุ่ม 0 น้อยกว่า?"] = compare["label 0"] < compare["label 1"]
compare
```

### 3.3 สัดส่วน label 0 ตามจำนวนครั้งที่เติมเงินใน 30 วัน

ถ้า 0 คือลูกค้าไม่ดีจริง คนที่ไม่ค่อยเติมเงินควรมีสัดส่วน label 0 สูง และคนที่เติมเงินบ่อยควรมีสัดส่วนต่ำ

```python
groups = pd.cut(df["cnt_ma_rech30"], bins=[-1, 0, 1, 4, np.inf], labels=["0 ครั้ง", "1 ครั้ง", "2-4 ครั้ง", "5+ ครั้ง"])
by_rech = df.groupby(groups, observed=True)["label"].agg(rows="size", pct_label0=lambda s: round((s == 0).mean() * 100, 1))
by_rech
```

## ขั้นที่ 4: Bad rate

bad rate = จำนวนลูกค้าไม่ดี (`label` = 0) ÷ จำนวนทั้งหมด

แหล่งข้อมูลบอกไว้ประมาณ 12.5% ต้องคำนวณเองแล้วเทียบ และดูว่าแต่ละช่วงเวลาคงที่ไหม

### 4.1 Bad rate รวม

```python
df["bad"] = (df["label"] == 0).astype(int)   # 1 = bad ตามแบบที่ใช้กันทั่วไปในงาน credit scoring
n_bad = df["bad"].sum()
print(f"bad = {n_bad:,} จาก {len(df):,} แถว")
print(f"bad rate = {df['bad'].mean():.2%}")
```

### 4.2 Bad rate แยกตามเดือน

```python
df["month"] = pd.to_datetime(df["pdate"], format="%d-%m-%Y").dt.to_period("M")
by_month = df.groupby("month")["bad"].agg(rows="size", n_bad="sum", bad_rate="mean")
by_month["bad_rate"] = by_month["bad_rate"].mul(100).round(2)
by_month
```

### 4.3 แถวที่ไม่ได้กู้เลย

ลูกค้าที่ไม่ได้กู้ (`cnt_loans30` = 0) ไม่มีเงินกู้ให้ผิดนัด จึงไม่ควรนับเป็น "ลูกค้าดี" 

```python
no_loan = df["cnt_loans30"] == 0
print(f"แถวที่ cnt_loans30 = 0: {no_loan.sum():,} แถว, bad rate = {df.loc[no_loan, 'bad'].mean():.2%}")
```

### 4.4 Bad rate ของกลุ่มที่ควรใช้พัฒนาโมเดล

เลือกเฉพาะ: เดือน มิ.ย.–ก.ค. 2016, มีการกู้ใน 30 วัน และลบแถวที่ซ้ำ

```python
dev = df[df["month"].astype(str).isin(["2016-06", "2016-07"]) & ~no_loan].drop_duplicates()
print(f"dev = {len(dev):,} แถว, bad = {dev['bad'].sum():,}, bad rate = {dev['bad'].mean():.2%}")
```
